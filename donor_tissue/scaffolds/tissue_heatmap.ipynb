```python
import altair as alt
import pandas as pd
from vega_datasets import data

# Fill in
font="Arial" # Change font here
top_count=50 # Change max count here
data = pd.DataFrame({
                        'Blood (3A)': [4, 6, 2, 38, 10],
                        'Frontal Lobe (3AK)': [5, 7, 3, 9, 11],
                        'Cerebellum (3AM)': [8, 5, 46, 4, 260],
                        'Asc. Colon (3E)': [14, 20, 7, 5, 30],
                        'Desc. Colon (3G)': [9, 11, 13, 15, 17]},
                        index=['Blood (3A)', 'Frontal Lobe (3AK)', 'Cerebellum (3AM)', 'Asc. Colon (3E)', 'Desc. Colon (3G)'])



heatmap_data = data.reset_index().melt(
    id_vars='index',
    var_name='column',
    value_name='value'
).rename(columns={'index': 'row'})

# Configure heatmap
heatmap = alt.Chart(heatmap_data).mark_rect().encode(
    x=alt.X('column:N', title=None, 
    sort=['Blood (3A)', 'Frontal Lobe (3AK)', 'Cerebellum (3AM)', 'Asc. Colon (3E)', 'Desc. Colon (3G)'],
    axis=alt.Axis(labelAngle=-45, labelFontSize=16, labelFont=font)),
    y=alt.Y('row:N', title=None, 
    sort=['Blood (3A)', 'Frontal Lobe (3AK)', 'Cerebellum (3AM)', 'Asc. Colon (3E)', 'Desc. Colon (3G)'],
    axis=alt.Axis(labelFontSize=16, labelFont=font)),
    color=alt.Color('value:Q', scale=alt.Scale(domain=[0, top_count, 260], range=["#eff3ff", "#08519c", "#08519c"]),
            legend=alt.Legend(title="Count", labelFontSize=16, labelFont=font, titleFont=font, titleFontSize=20)), # Change custom color scale here if desired
    tooltip=['row:N', 'column:N', 'value:Q']
).properties(
    title=alt.Title('SMHT005 Tissue Intersection Counts', fontSize=28, anchor='middle', font=font),
    width=600,
    height=600,
)

# Configure conditional text color based on cell value (top_count/2 threshold)
color = (
    alt.when(alt.datum.value < top_count / 2)
    .then(alt.value("black"))
    .otherwise(alt.value("white"))
)
# Configure text
text = heatmap.mark_text(baseline='middle', size=24, font=font).encode(
    text=alt.Text('value:Q'), color=color
)

# Draw the chart
heatmap + text
```

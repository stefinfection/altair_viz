```python
import altair as alt
import pandas as pd
import numpy as np

hash_counts = [268,
100,
84,
39,
32,
24,
21,
23,
18,
26,
13,
12,
13,
9,
17,
21,
12,
10,
17,
7,
17,
12,
18,
7,
13,
12,
10,
10,
12,
#685
]
hash_total=sum(hash_counts)

paired_control_counts = [387,
168,
161,
118,
119,
94,
90,
91,
79,
72,
85,
68,
80,
71,
71,
62,
65,
62,
54,
48,
69,
70,
60,
50,
51,
62,
51,
55,
56,
#10880
]
paired_control_total=sum(paired_control_counts)

overlapping_counts =[
167,
187,
157,
152,
124,
151,
123,
133,
115,
93,
110,
111,
93,
115,
92,
87,
94,
102,
92,
90,
100,
104,
117,
113,
119,
114,
121,
117,
115,
#38053
]
overlapping_total=sum(overlapping_counts)

# Your color palette
red = '#D79DB8'
orange = '#F5D5A4'
green = '#A7D8D9'
blue = '#4D90B6'
purple = '#9D8BA7'

# Create the frequency intervals
bin_labels = [f"{i*1}-{(i+1)*1}%" for i in range(29)]
#bin_labels.append("29+%")  # Last bin is 29-100%

# Combine all data into a single DataFrame
data_rows = []
degrees_data = {
    'Hash Counts': (hash_counts, blue),
    'Paired Ctrl Counts': (paired_control_counts, orange),
    'Overlapping Counts': (overlapping_counts, green)
}

for degree_name, (counts, color) in degrees_data.items():
    for i, count in enumerate(counts):
        data_rows.append({
            'frequency_interval': bin_labels[i],
            'percentage': count/(sum(counts)) * 100,
            'degree': degree_name
        })

# Create DataFrame
combined_df = pd.DataFrame(data_rows)

# Create the grouped bar chart
chart = alt.Chart(combined_df).mark_bar(
    stroke='white',
    strokeWidth=1
).add_params(
    alt.selection_interval()
).encode(
    x=alt.X('frequency_interval:O',
            title='Allele Frequency Intervals',
            axis=alt.Axis(labelAngle=-45, labelFontSize=12, titleFontSize=16),
            sort=bin_labels),
    y=alt.Y('percentage:Q',
            title='Percentage',
            axis=alt.Axis(tickMinStep=1, labelFontSize=12, titleFontSize=16)),
    color=alt.Color('degree:N',
                    title='',
                    scale=alt.Scale(domain=['Hash Counts', 'Paired Ctrl Counts', 'Overlapping Counts'],
                                   range=[blue, orange, green]),
                                   legend=alt.Legend(titleFontSize=20, labelFontSize=12, symbolSize=100, symbolType='circle')),    xOffset=alt.XOffset('degree:N'),
    tooltip=['degree:N', 'frequency_interval:O', 'percentage:Q'],
).properties(
    width=800,
    height=400,
    title='Variant Allele Frequency Distribution by Percentage'
    ).resolve_scale(
    color='independent'
)

# Display the chart
chart.show()
```
